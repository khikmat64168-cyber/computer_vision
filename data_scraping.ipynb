```python
import os, requests
from io import BytesIO
from PIL import Image
from urllib.parse import quote_plus
from playwright.async_api import async_playwright

async def download_images(query, limit=100, output_dir="/content/images"):
    os.makedirs(output_dir, exist_ok=True)

    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox", "--disable-dev-shm-usage"]
        )
        page = await browser.new_page(locale="en-US")
        await page.goto(
            f"https://www.bing.com/images/search?q={quote_plus(query)}",
            wait_until="domcontentloaded",
            timeout=60000
        )

        for _ in range(8):
            await page.mouse.wheel(0, 1500)
            await page.wait_for_timeout(1500)

        results = await page.locator("a.iusc").evaluate_all("""
            els => els.map(e => {
                try { return JSON.parse(e.getAttribute('m')); }
                catch { return null; }
            }).filter(Boolean)
        """)
        await browser.close()

    count = len([
        f for f in os.listdir(output_dir)
        if f.lower().endswith(".jpg")
    ])
    seen = set()

    for item in results:
        if count >= limit:
            break

        url = item.get("murl")
        if not url or url in seen:
            continue
        seen.add(url)

        try:
            r = requests.get(url, timeout=15,
                             headers={"User-Agent": "Mozilla/5.0"})
            r.raise_for_status()
            img = Image.open(BytesIO(r.content)).convert("RGB")

            if min(img.size) < 200:
                continue

            count += 1
            img.save(f"{output_dir}/image_{count}.jpg")
            print(f"Downloaded image_{count}.jpg")

        except Exception:
            continue

    print(f"Total images: {count}")
```

```python

await download_images(
    "uzbek pilaf plov food",
    limit=100,
    output_dir="/content/pilaf_images"
)
```

```python

import os, shutil

source_dir = '/content/pilaf_images'
target_dir = '/content/drive/MyDrive/food101_dataset/pilaf '

os.makedirs(target_dir, exist_ok=True)

for filename in os.listdir(source_dir):
  src_file = os.path.join(source_dir, filename)
  dst_file = os.path.join(target_dir, filename)
  shutil.move(src_file, dst_file)
```
